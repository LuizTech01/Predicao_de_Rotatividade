```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dt = pd.read_csv('../dataset/WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
# Converte para string e limpa espaços/maiúsculas
dt['Churn'] = dt['Churn'].astype(str).str.strip().str.capitalize()

# Aplica o mapeamento
dt['Churn'] = dt['Churn'].map({'Yes': 1, 'No': 0})

correlation_matrix = dt.corr(numeric_only=True)

plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Matriz de Correlação")
plt.show()
```

```python
dt.to_excel('../dataset/WA_Fn-UseC_-Telco-Customer-Churn.xlsx', index=False)
```
